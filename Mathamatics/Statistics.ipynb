### **Statistics**

#### **Mean (Average)**

```python
delivery_days = [2, 3, 4, 5, 6]

total = sum(delivery_days)

count = len(delivery_days)

mean = total / count

print("Average delivery days:", mean)
```

How Machine Learning uses Mean
- Predicting average house prices.
- Understanding average customer spending.
- Analyzing average delivery time.
- Filling missing values in datasets.
- Understanding data before training a model.

### **Project: E-commerce Data Statistics Analyzer**

```python
import pandas as pd

# Create dataset
data = {
    "delivery_days": [2, 3, 4, 5, 20],
    "order_amount": [500, 700, 800, 900, 5000],
    "category": ["Mobile", "Laptop", "Mobile",
                 "Tablet", "Mobile"]
}

df = pd.DataFrame(data)

# Calculate statistics
print("Mean:")
print(df["delivery_days"].mean())

print("Median:")
print(df["delivery_days"].median())

print("Mode:")
print(df["category"].mode()[0])
```

### **Range in Statistics**

### **Standard Deviation**

Standard Deviation tells us how far data values typically spread from their average.

Imagine an e-commerce company tracking delivery times.
- Average delivery time = 5 days.
- Some orders arrive in 4 days.
- Some orders arrive in 5 days.
- Some orders arrive in 6 days.
Standard Deviation helps the company understand how much delivery times usually differ from 5 days.

### **Understand it visually**

```python
delivery_days = [4, 5, 6]
```

```python
import statistics

delivery_days = [4, 5, 6]

# Calculate average
average = statistics.mean(delivery_days)

# Calculate standard deviation
sd = statistics.pstdev(delivery_days)

print("Average:", average)
print("Standard Deviation:", sd)
```

### **Percentiles**

### **Correlation**

###  **What is Correlation?**
Correlation tells us how two variables are related to each other.

In simple English:

When one value changes, correlation helps us understand how another value tends to change.

For example:

Imagine a company wants to understand the relationship between:
- Advertising budget
- Number of products sold
If advertising spending increases and product sales also increase, they may have a positive correlation.

```python
import pandas as pd

# Create dataset
data = {
    "house_size": [500, 800, 1000, 1500, 2000],
    "bedrooms": [1, 2, 2, 3, 4],
    "house_price": [20, 35, 45, 70, 90]
}

df = pd.DataFrame(data)

# Calculate correlation
correlation = df.corr()

print(correlation)
```

```python
print(df["house_size"].corr(df["house_price"]))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns


# Plot heatmap
plt.figure(figsize=(8, 5))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("House Price Correlation Heatmap")
plt.show()
```

### **Outliers**

### **What is an outlier?**
An outlier is a data value that is unusually different from most other values in a dataset.

Imagine you're working for an IT company. You collect the monthly salaries of 10 employees.

```python
salaries = [20000, 22000, 21000, 25000, 23000,
            24000, 22000, 21000, 23000, 200000]
```

```python
import numpy as np

salaries = np.array([
    20000, 22000, 21000, 25000, 23000,
    24000, 22000, 21000, 23000, 200000
])

print("Minimum:", np.min(salaries))
print("Maximum:", np.max(salaries))
print("Mean:", np.mean(salaries))
print("Median:", np.median(salaries))
```
