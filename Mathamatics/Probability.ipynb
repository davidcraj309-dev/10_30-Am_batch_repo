Let's calculate the probability of getting an even number on a dice.

```python
total_outcomes = 6

favourable_outcomes = 3

probability = favourable_outcomes / total_outcomes

print("Probability:", probability)
print("Percentage:", probability * 100)
```

```python
source_a = 0.60
source_b = 0.40

spam_a = 0.10
spam_b = 0.20

total_spam = (spam_a * source_a) + (spam_b * source_b)

print(total_spam)
```

Notice something important:

Even though 90% of spam emails contain "FREE", an email containing "FREE" has only a 69.23% probability of being spam in this example.

That's why understanding prior probability matters!

```python
spam = 0.20
not_spam = 0.80

free_given_spam = 0.90
free_given_not_spam = 0.10

free_probability = (
    free_given_spam * spam
    + free_given_not_spam * not_spam
)

spam_given_free = (
    free_given_spam * spam
    / free_probability
)

print(spam_given_free)
```
