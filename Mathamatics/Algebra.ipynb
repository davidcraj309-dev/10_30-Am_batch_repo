### **Real-world Machine Learning problem**

Suppose you want to predict house prices based on house size.

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
```

```python
data = {
    'House_Size': [500, 600, 700, 800, 900,
                   1000, 1100, 1200, 1300, 1400],

    'House_Price': [25, 30, 35, 40, 45,
                    50, 55, 60, 65, 70]
}

df = pd.DataFrame(data)

print(df)
```

```python
X = df[['House_Size']]
X
```

```python

y = df['House_Price']
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
```

```python
model = LinearRegression()

model.fit(X_train, y_train)
```

```python
predictions = model.predict(X_test)

print(predictions)
```

```python
new_house = pd.DataFrame({
    'House_Size': [1200]
})

predicted_price = model.predict(new_house)

print("Predicted House Price:", predicted_price[0], "Lakhs")
```

```python
plt.scatter(
    df['House_Size'],
    df['House_Price']
)

plt.xlabel("House Size (Sq.ft)")
plt.ylabel("House Price (Lakhs)")

plt.title("House Size vs House Price")

plt.show()
```

```python
# 4. Predict prices
y_pred = model.predict(X)

# 5. Plot training data
plt.scatter(
    X, y,
    marker='x',
    label='Training Data'
)

# 6. Plot regression line
plt.plot(
    X, y_pred,
    color='red',
    label='Linear Regression Line'
)

# 7. Add labels and title
plt.xlabel("House Size (sqft)")
plt.ylabel("Price (in $1000s)")

plt.title("Linear Regression Fit for House Price Prediction")

# 8. Show legend and grid
plt.legend()
plt.grid(True)

# 9. Display graph
plt.show()
```

### **How does Machine Learning use X and y?**

Think of a Machine Learning model as a student.

You give the student previous examples:
- 500 sq.ft → ₹25 lakhs
- 700 sq.ft → ₹35 lakhs
- 900 sq.ft → ₹45 lakhs
The student learns the relationship between house size and house price.

Every time the house size increases by 200 sq.ft, the price increases by ₹10 lakhs.

Later, you ask:

"What will be the price of a 1,500 sq.ft house?"

The model uses what it learned to predict the answer.

This is called Supervised Learning.

### **E-commerce sales prediction**

Imagine you work for an e-commerce company.

The company wants to understand how advertising expenditure affects product sales.

```python
from sklearn.linear_model import LinearRegression
```

```python
# Advertising budget
X = [[1000], [2000], [3000], [4000], [5000]]

# Products sold
y = [20, 40, 60, 80, 100]
```

```python
model = LinearRegression()
model.fit(X, y)
```

### **Extract slope and intercept**

```python
m = model.coef_[0]

c = model.intercept_

print("Slope:", m)
print("Intercept:", c)
```

```python
prediction = model.predict([[6000]])

print("Predicted Sales:", prediction[0])
```

### **Functions and graphs**

### **What is a function?**
A function is simply a relationship between an input and an output.

Imagine you own an e-commerce business.

A customer visits your online store, and you want to predict how many products will be sold based on advertising spending.

You provide an advertising budget, and a mathematical function calculates the predicted sales.

### **What is a graph?**
A graph is a visual representation of the relationship between two variables.

In our e-commerce project:
- X-axis = Advertising budget.
- Y-axis = Predicted sales.

```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

X = [[0], [1000], [2000], [3000], [4000]]

y = [10, 30, 50, 70, 90]

model = LinearRegression()

model.fit(X, y)

prediction = model.predict([[5000]])

print("Predicted Sales:", prediction[0])

plt.scatter(X, y)
plt.plot(X, model.predict(X))

plt.xlabel("Advertising Budget")
plt.ylabel("Sales")
plt.title("Linear Regression Graph")

plt.show()
```
