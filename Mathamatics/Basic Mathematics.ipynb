### **1. Probability – Student Pass Prediction**

Imagine you are building a Machine Learning model to predict whether students will pass or fail.

Your dataset contains 100 students.
- 80 students passed.
- 20 students failed.

### **Step 1: Calculate the fraction**

### **How ML uses this:**

A classification model can predict that a student has an 80% probability of passing.

```python
total_students = 100
passed_students = 80

probability = passed_students / total_students

print(probability)
```

### **2. Model Accuracy – Measuring ML performance**

### **Imagine a school scenario**
You are building a Machine Learning model using a Decision Tree.

You give your model 100 students to predict.

After checking the actual results, you get:
- 90 students: Model predicted correctly.

```python
actual = 100
correct = 90

accuracy = correct / actual

print("Accuracy:", accuracy * 100, "%")
```

### **Train-Test Split – Dividing datasets**

Imagine your e-commerce dataset contains 2,000 records.

You decide:
- 80% training data.
- 20% testing data.

```python
import pandas as pd
from sklearn.model_selection import train_test_split # divides your dataset into training and testing data.

data = {
    'Study_Hours': [1, 2, 3, 4, 5, 6, 7, 8],
    'Attendance': [40, 50, 60, 65, 70, 75, 80, 90],
    'Result': ['Fail', 'Fail', 'Fail', 'Pass',
               'Pass', 'Pass', 'Pass', 'Pass']
}

df = pd.DataFrame(data)
df
```

```python
# Input features
X = df[['Study_Hours', 'Attendance']] #Because we are selecting multiple columns from the DataFrame.
X
```

```python
# Target variable
y = df['Result']
y
```

```python
# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2
)

print(y_train)
print(y_test)
```

```python
print("Total Records:", len(df))

print("Training Records:", len(X_train))

print("Testing Records:", len(X_test))
```

Here:
- Training = 80%
- Testing = 20%
- Ratio = 4:1
ML application: Dividing datasets into training and testing portions.

```python
print("X Training Data:")
print(X_train)

print("X Testing Data:")
print(X_test)
```

```python
print("Y Training Data:")
print(y_train)

print("Y Testing Data:")
print(y_test)
```

### **Lesson 3: Ratios and Proportions**

### **Part 1: What is a ratio?**

A ratio means comparing two quantities.

Imagine you are in a classroom.

There are:
- 10 boys
- 5 girls
How do we compare boys and girls?

### **Part 2: What is Proportion?**

A proportion means two ratios are equal.

Let's understand with a real-world example.

Imagine you buy:
- 2 apples = ₹20
- 4 apples = ₹40

### **What is Inverse Proportion?**

Inverse means opposite.

Proportion means relationship between two quantities.

Inverse Proportion means:
- When one quantity increases, the other quantity decreases.
- When one quantity decreases, the other quantity increases.
Both move in opposite directions.

### **Real-world example: Temperature prediction**

Imagine a weather prediction system.

#### **Very important: Positive and negative numbers in Machine Learning**

```python
import pandas as pd

data = {
    "Day": [1, 2, 3, 4, 5],
    "Temperature": [35, 20, 5, -5, -10]
}

df = pd.DataFrame(data)

print(df)
```

```python
X = df[["Temperature"]]
```

```python
from sklearn.preprocessing import StandardScaler

# Create StandardScaler
scaler = StandardScaler()
```

In StandardScaler:
- fit() calculates the mean and standard deviation.
- transform() uses them to standardize the values.

```python
# Transform the data
X_scaled = scaler.fit_transform(X)

print(X_scaled)
```
