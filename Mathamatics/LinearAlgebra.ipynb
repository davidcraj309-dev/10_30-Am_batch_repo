### **Linear Algebra**

Why is Linear Algebra important in Machine Learning?

Machine Learning models work with numbers. Images, text, customer information, prices, and other data are represented as numbers.

Linear Algebra helps us organize, calculate, and process these numbers efficiently.

```python
age = 25
salary = 25000
temperature = 32

print(age)
print(salary)
print(temperature)
```

Where is a vector used in ML?

* Remember: Vector = Collection of numbers in an ordered arrangement.

```python
import numpy as np

employees = np.array([
    [25, 25000, 2],
    [28, 30000, 4],
    [26, 28000, 3]
])

print(employees)
print(employees.shape)
```

```python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

result = np.matmul(A, B)

print(result)
```

Where is it used in Machine Learning?
- Neural networks
- Deep learning
- Image recognition
- Prediction models
- Processing large datasets
For example, a neural network uses matrix multiplication to calculate values as information moves through its layers.

Remember: Matrix multiplication = Row × Column.

```python
import numpy as np

A = np.array([2, 3, 4])
B = np.array([5, 6, 7])

result = np.dot(A, B)

print(result)
```
