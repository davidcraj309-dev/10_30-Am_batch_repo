### **1. Reverse a string**

```python
text =  "python"

def reverse_str(text):

    reversed = ""
    for each in text:
      reversed = each+reversed
    return reversed

ans = reverse_str(text)
print(ans)
```

### **2. Check if a string is palindrome**

```python
text = "madam"

def reverse_str(text):

    reversed = ""

    for each in text:
        reversed = each + reversed

    if text == reversed:
        return "is palindrome"
    else:
        return "not palindrome"

ans = reverse_str(text)
print(ans)
```

### **7. Count frequency of elements**

```python
numbers = [10, 20, 10, 30, 20, 10]

def count_frequency(numbers):

    frequeny = {}

    for each in numbers:
        if each in frequeny:
            frequeny[each] += 1
        else:
            frequeny[each] = 1

    return frequeny

ans = count_frequency(numbers)
print(ans)
```

### **10. Find second largest number**

```python
numbers = [10, 20, 30, 40, 50]

def second_largest(numbers):
    numbers = list(set(numbers))
    numbers.sort()
    return numbers[-2]


result = second_largest(numbers)
print(result)
```

### **16.Find duplicate values**


```python
numbers = [1, 2, 3, 2, 4, 1, 5]


def find_duplicate(numbers):
    seen = []
    duplicate = []
    for each in numbers:
        if each in seen:
            duplicate.append(each)
        else:
            seen.append(each)

    return duplicate

ans = find_duplicate(numbers)
print(ans)
```

### **17.Factorial**

```python
number = 5

def factorial(number):
    result = 1

    for i in range(1, number + 1):
        result *= i

    return result

answer = factorial(number)
print(answer)
```

### **18.Fibonacci series**

```python
def fibonacci(n):
    a = 0
    b = 1
    result = []

    for i in range(n):
        result.append(a)
        a, b = b, a + b

    return result

answer = fibonacci(10)
print(answer)
```

### **19. Check prime number**

```python
number = 7

is_prime = True

if number < 2:
    is_prime = False
else:
    for i in range(2, number):
        if number % i == 0:
            is_prime = False
            break

if is_prime:
    print("Prime")
else:
    print("Not Prime")
```
