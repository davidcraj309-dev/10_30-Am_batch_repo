### **1.What is the difference between syntax errors and runtime errors in Python?**

* Syntax error - Error in the way you write the code. Python cannot understand the code.
* Runtime error - Error that happens while the program is running.

#### **Syntax error Example**

```python
print("Hello" # Syntax error
```

#### **Runtime error Example**

```python
x = 10 / 0 # ZeroDivisionError: division by zero
print(x)
```

### **2.Explain the role of the try, except, else, and finally blocks.**

* try - Contains code that may cause an error
* except - Handles the error if it occurs.
* else - Runs when there is no error in the try block.
* finally - Runs whether an error occurs or not.

```python
try:
    print("Welcome to Zomato")
    number_of_items = int(input("How many items? "))
    total_price = 200 * number_of_items
    average_price = total_price / number_of_items
    print("Average price per item:", average_price)

except ZeroDivisionError:
    print("Cannot divide by zero")

else:
    print("No error found")

finally:
    print("Code always executes")

print("Next code block")
```

### **3.What happens if an exception is raised but not handled?**

* If an exception happens and we don't handle it, the program stops and shows an error message.

#### **If an exception is raised but not handled:**

```python
x = 10 / 0
print("Next code")  # Next code will not execute if the exception is not handled.
```

### **4.What is the difference between Exception and BaseException in Python?**
* BaseException → Top-level base class for exceptions.
* Exception → Used for common errors like ValueError, TypeError, etc.

```python
try:
    number = int(input("Enter a number: "))
    result = 10 / number
    print(result)

except Exception as e:
    print("Exception handled:", e)
```

### **5.Explain the use of the raise keyword with an example.**

```python
valid_id = 123

id_number = int(input("Enter your ID: "))

try:
    if id_number != valid_id:
        raise ValueError("Invalid ID")

    print("ID verified. Entry allowed.")

except ValueError as e:
    print(e)
    print("Entry denied.")
```

### **6.Explain the use of the raise keyword with an example.**

```python
age = int(input("Enter age: "))

try:
    if age < 18:
        raise ValueError("Age must be 18 or above")

    print("Eligible")

except ValueError as e:
    print(e)
```

### **7.What is the purpose of the assert statement, and how does it differ from raising an exception?**

* assert - checks an assumption.
* raise - intentionally creates an exception for a specific situation.

```python
age = 20

assert age >= 18, "User must be 18 or above"

print("Registration successful")
```

```python
age = 15

if age < 18:
    raise ValueError("User must be 18 or above")

print("Registration successful")
```

### **8.Can we use multiple except blocks in Python? Give an example.**


```python
try:
    amount = int(input("Enter withdrawal amount: "))

    balance = 5000

    if amount > balance:
        raise ValueError("Insufficient balance")

    print(f"Please collect your cash {amount}")

except ValueError as e:
    print("Error:", e)

except TypeError:
    print("Invalid data type")

except Exception:
    print("Something went wrong")
```
