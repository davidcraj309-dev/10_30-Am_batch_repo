### 1. Reverse a String

```python
def reverseStr(str):
    result = ""
    each = ""
    for each in str:
        result = each + result
         
    return result

ans = reverseStr("Python")
print(ans)
```

### 2. Count Characters

```python
def counteChr(str):
    count = 0
    for each in str:
           count += 1
         
    return count

ans = counteChr("Python")
print(ans)
```

### 3. Count Vowels

```python
def countVowel(str):
    count = 0
    for each in str:
        if each in "aeiou":
           count += 1
         
    return count

ans = countVowel("Python")
print(ans)
```

### 4. Count Consonants

```python
def countConst(str):
    count = 0
    for each in str:
        if each not in "aeiou":
           count += 1
         
    return count

ans = countConst("Python")
print(ans)
```

### 5. Count Digits

```python
def countDigit(str):
    count = 0
    for each in str:
        if each>"0" and each<="9":
           count += 1
         
    return count

ans = countDigit("David12345")
print(ans)
```

### 6. Count Spaces

```python
def counteSpace(str):
    count = 0
    for each in str:
        if each  == " ":
           count += 1
         
    return count

ans = counteSpace("David Raj")
print(ans)
```

### 7. Count Uppercase Letters

```python
def countUppercase(str):
    count = 0

    for each in str:
        if each >= "A" and each <= "Z":
            count += 1

    return count


ans = countUppercase("DaViD Raj")
print(ans)
```

### 8. Count Lowercase Letters

```python
def countCharacter(str):
    count = 0

    for each in str:
        if each >= "a" and each <= "z":
            count += 1

    return count


ans = countUppercase("DaViD Raj")
print(ans)
```

### 9. Count a Specific Character

```python
def countCharacter(text, target):
    count = 0

    for each in text:
        if each == target:
         count += 1

    return count

ans = countCharacter("banana", "a")
print(ans)
```

### 10. Find First Position of a Character

```python
def countCharacter(text, target):
    count = 0

    for each in text:
        if each == target:
         count += 1

    return count

ans = countCharacter("python", "t")
print(ans)
```

### 10. Count Zeros

```python
def countZero(zero):
    count = 0

    for each in zero:
        if each == "0":
         count += 1

    return count

ans = countZero("100200300")
print(ans)
```

### 10. Remove Spaces

```python
def removespace(space):
    newstr = ""

    for each in space:
        if each == " ":
         continue
        newstr = newstr + each

    return newstr

ans = removespace("David Raj Python")
print(ans)
```
