###  **Task 2:**  **Nested while loop Task**


```python
row = int(input("Enter a row number: "))
column = int(input("Enter a column number: "))


i = 1

while row>=i:
    j =  1

    while column>=j:
        print(j,end="")
        j += 1

    i += 1
    print()
```
