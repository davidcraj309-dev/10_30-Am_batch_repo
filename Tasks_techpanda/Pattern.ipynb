### **1.Right-aligned triangle:**

```python
num = int(input("Enter number of rows: "))



for i in range(num):
    for j in range(num-1-i):
        print(" " , end="")

    for k in range(i+1):
        print("*",end="")

    print()
```

### **2.Reverse right-aligned:**

```python
num = int(input("Enter number of rows: "))


for i in range(num):
    for j in range(i+1):
        print(" ",end="")

    for k in range(num-i):
        print("*",end="")

    print()
```

### **3.Pyramid:**

```python
num = int(input("Enter number of rows: "))

for i in range(num):
    for j in range(num-1-i):
        print(" ",end="")

    for k in range(2*i+1):
        print("*",end="")

    print()
     
```

### **4.Reverse pyramid:**

```python
num = int(input("Enter number of rows: "))

for i in range(num):

    for j in range(i):
        print(" ", end="")

    for k in range(2 * (num - i) - 1):
        print("*", end="")

    print()
```
