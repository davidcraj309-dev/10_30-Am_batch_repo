### **1.Remove Duplicate Values**

```python
input = [10, 30, 55, "hello", 30, 55, 10, True, "Hello", "hello", False, 10, True, 30, 55, "Hello"]

def remove_duplicate(input):

    original = []
    
    
    for each in input:
        if each not in original:
            original.append(each)

    return original   
  
ans = remove_duplicate(input)
print(ans)
```
