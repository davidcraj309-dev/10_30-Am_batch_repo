#### *92.Create an ATM withdrawal program.*

#### *105.Encapsulation Challenge*

````python
class Bankaccount:

    def __init__(self, account_holder, account_number, balance):
        self.a = account_holder
        self.b = account_number
        self.__c = balance

    def display_account(self):
        print(f"Account holder: {self.a}")
        print(f"Account number: {self.b}")

    def deposit(self, amount):
        self.__c += amount
        print(f"Deposit amount: {amount}")

    def withdraw(self, amount):
        self.__c -= amount
        print(f"Withdraw amount: {amount}")

    def get_balance(self):
        print(f"Current balance: {self.__c}")


david = Bankaccount("David Chinnappa Raj", 292388399494, 20000)

david.display_account()
david.deposit(5000)
david.withdraw(2000)
david.get_balance()
````

#### *111.Decorator with *args*

````python
def adding_num(func):
    def inner(*args):
        print("Adding numbers...")
        result = func(*args)
        print(f"Result:{result}")
        
    return inner

@adding_num
def add(a,b):
    return a+b

add(10,10)
````

#### *112.Authentication decorator*

````python
def login_required(func):
    logged_in = True
    def inner_login(amount):
        if logged_in == True:
            func(amount)

    return inner_login

@login_required
def withdraw_money(amount):
    print(f"your withdraw amount: {amount}")

withdraw_money(200)
````

#### *114.Input validation decorator*

````python
def validate_prompt(func):

    def inner_calling(prompt):
        if prompt == "":
            print(" cannot be empty")
        else:
            func(prompt)

    return inner_calling


@validate_prompt
def ask_ai(prompt):
    print("AI:", prompt)

ask_ai("Explain RAG")
````
