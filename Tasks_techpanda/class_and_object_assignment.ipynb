#### *3.Create a class Rectangle with attributes length and width. Add methods to calculate area and perimeter.*

```python
class Rectangle:
    length = 100
    width = 50

    def calculate_area(self):
        area = self.length*self.width
        print(f"area:{area}")
       
    def perimeter(self):
        perimeter = self.length + self.width + self.length + self.width
        print(f"perimeter:{perimeter}")

obj = Rectangle()
obj.calculate_area()
obj.perimeter()
```

#### *5.Define a class Circle with a class variable pi = 3.14. Add a method to calculate the area.*

```python
class Circle:
    pi = 3.14

    def Calculate_area(self,radius):
        area = (self.pi * radius*radius) # formula A=πr2
        print(f"area:{area}")

obj = Circle()
obj.Calculate_area(5)
```

#### *6.Create a class BankAccount with attributes account_number, name, and balance. Add methods to deposit, withdraw, and display balance.*

```python
class BankAccount:

    def __init__(self,account_number,name,balance):
        self.account_number = account_number
        self.name = name
        self.balance = balance

    def deposit(self,amount):
        self.balance  = self.balance + amount
        print(f"deposit_amount:{amount}")
        print(f"total_balance:{self.balance}")

    def withdraw(self,amount):
        self.balance  = self.balance - amount
        print(f"withdraw_amount:{amount}")
        print(f"total_balance:{self.balance}")

    def display_balance(self):
        print(f"final_balance:{self.balance}")


obj = BankAccount("348748t75498","david",500)
obj.deposit(100)
print("-----------")
obj.withdraw(200)
print("-----------")
obj.display_balance()
```

#### *7.Write a class Employee with attributes emp_id, name, and salary. Add a method to give a 10% bonus to employees.*

```python
class Employee:

    def __init__(self,emp_id,name,salary):
        self.emp_id = emp_id
        self.name = name
        self.salary = salary      

    def employee_bonus(self,bonus):
        bonus  = self.salary * bonus/100
        self.salary = self.salary + bonus
        print(self.salary)

    def employee_details(self):
        print(f"emp_id:{self.emp_id}")
        print(f"name:{self.name}")
        print(f"salary:{self.salary}")

obj = Employee(123,"david",500)
obj.employee_bonus(10)
obj.employee_details()
```

#### *8.Implement a class Library with methods to add_book(title), remove_book(title), and display_books().*

```python
class Library:
    books = []

    def add_book(self,title): 
        self.books.append(title)
        print(f"Books_added:{self.books}")

    def remove_book(self,title):
        self.books.remove(title)
        print(f"Books_removed:{self.books}")

    def display_books(self):
        print(f"All_Books:{self.books}")

obj = Library()
obj.add_book("Python")
obj.add_book("Java")
obj.remove_book("Python")
obj.display_books()
     

```

#### *9.Create a class Calculator with methods add, subtract, multiply, and divide. Demonstrate with objects.*

```python
class Calculator:

    def add(self,a,b):
        addition = a + b
        print(f"Addition:{addition}")

    
    def subtract(self,a,b):
        subtraction = a - b
        print(f"subtraction:{subtraction}")

    
    def multiply(self,a,b):
        Multiplication  = a * b
        print(f"Multiplication :{Multiplication}")

    def divide(self,a,b):
        division  = a / b
        print(f"division :{division}")

obj = Calculator()
obj.add(2,2)
obj.subtract(2,2)
obj.multiply(2,2)
obj.divide(2,2)

```

#### *10.Write a program to show the difference between instance variables and class variables with an example*

```python
class person:
    #class variables
    course = "GenAi"

    def __init__(self,name,age,city): 
        ### Instance variables
        self.name = name
        self.age = age
        self.city = city

    def display_person(self):
        print(f"name:{self.name}")
        print(f"age:{self.age}")
        print(f"city:{self.city}")
        print(f"course:{self.course}")

obj = person("david",27,"chennai")
obj.display_person()
```
