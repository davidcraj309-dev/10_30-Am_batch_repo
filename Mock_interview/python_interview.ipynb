### **C. Looping Programs**

#### *1.Write a program to extract uppercase letters, lowercase letters, digits, and special characters into separate strings from a given string.*

````python
upper_case = []
lower_case = []
digits = []
special_char = []

text = input("inter Enter a string : ") #David@123

for char in text:
    if char.isupper():
        upper_case.append(char)
    elif char.islower():
        lower_case.append(char)
    elif char.isdigit():
        digits.append(char)
    else:
        special_char.append(char)


print(upper_case)
print(lower_case)
print(digits)
print(special_char)
````

## Factorial Table

We use the symbol `!` to represent factorial.

For example:

$$
5! = 5 \times 4 \times 3 \times 2 \times 1 = 120
$$

| Number | Factorial | Answer |
|---:|---|---:|
| 1 | $1!$ | 1 |
| 2 | $2 \times 1$ | 2 |
| 3 | $3 \times 2 \times 1$ | 6 |
| 4 | $4 \times 3 \times 2 \times 1$ | 24 |
| 5 | $5 \times 4 \times 3 \times 2 \times 1$ | 120 |

#### *2.Write a program to find the factorial of a given number*

````python
number = int(input("Enter a Number"))

def factorial(number):
    result = 1

    for i in range(1,number+1):
        result = result*i

    return result

find_factorial = factorial(number)
print(find_factorial)
````

### String Reversal — Step-by-Step Execution

**Input:** `hello`

| Loop | each | reversed_txt after update |
|---:|:---:|:---|
| 1 | `h` | `h` |
| 2 | `e` | `eh` |
| 3 | `l` | `leh` |
| 4 | `l` | `lleh` |
| 5 | `o` | `olleh` |

**Final reversed string:** `olleh`

#### *4. Write a program to reverse a given string*

````python
text = input("Enter a text: ")

def reverse_txt(text):
    reversed_txt = "" #olleh

    for each in text:
        reversed_txt = each + reversed_txt  #olleh

    return reversed_txt


result = reverse_txt(text)
print(result)
````

### **B. Conditional Statement Programs**

#### *8.Write a program to find the second-smallest number among a given set of numbers*

````python

numbers = [10, 5, 8, 2, 15]

smallest = numbers[0] #10
second = numbers[0] #10

for each in numbers:
    if each < smallest: #10<10=False,#5<10=True
        second = smallest #second = 10
        smallest = each #smallest =  5
    elif each < second and each != smallest: #comes here checks condition,10<10 and 10!=10 = False
        second = each

print("Second smallest number:", second)
````

| Step | Comparison | Result | What happens? | `smallest` |
|---|---|---|---|---:|
| Start | Input: a=10, b=5, c=8, d=2, e=6 | — | Five numbers are stored in variables | — |
| 1 | `a < b` → 10 < 5 | False | Skip the first `if` block and enter the outer `else` | — |
| 2 | `b < c` → 5 < 8 | True | Execute `smallest = b` | 5 |
| 3 | `d < smallest` → 2 < 5 | True | Execute `smallest = d` | 2 |
| 4 | `e < smallest` → 6 < 2 | False | Skip `smallest = e`; value stays the same | 2 |
| Final | `print("Smallest:", smallest)` | — | Display the smallest number | 2 |

### *10.Write a program to find the second-greatest number among four numbers*

````python

a = int(input("A: "))
b = int(input("B: "))
c = int(input("C: "))
d = int(input("D: "))

numbers = [a, b, c, d]

greatest = numbers[0]
second = numbers[0]

for each in numbers:
    if each > greatest:
        second = greatest
        greatest = each

    elif each > second and each != greatest:
        second = each

print("Second greatest:", second)
````

### *26. Write a program to check whether a given string is a palindrome.*

````python
text = "madam"

def reverse_str(text):

    reversed = ""

    for each in text:
        reversed = each + reversed

    if text == reversed:
        return "is palindrome"
    else:
        return "not palindrome"

ans = reverse_str(text)
print(ans)
````
