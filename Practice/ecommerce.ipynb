```python
import os

path = "C:\\Users\\Yesu\\OneDrive\\Documents\\Desktop"

file_name = "ecommerce.log.txt"


print(os.path.join(path,file_name))
```

```python
with open(os.path.join(path,file_name)) as my_log:
    content = my_log.read()
    print(content)
```

```python
with open(os.path.join(path,file_name)) as my_log:
    for each in my_log:
        print(each)
        break
```

```python
my_count = {}

with open(os.path.join(path,file_name)) as my_log:
    for each_line in my_log:
        status_code = each_line.split(" ")[1]

        if  status_code not in my_count:
            my_count[status_code] = 1
        else:
            my_count[status_code] += 1

print(my_count)  

        
```

```python
import csv 
with open("status_frequency.csv","w",newline="") as my_csv:
    writer = csv.writer(my_csv)

    writer.writerow(["Status_code","Count"])

    for status_code, count in my_count.items():
        writer.writerow([status_code,count])

print("Status frequency saved successfully!")
```

```python
fruits = ["apple", "banana", "apple", "orange",
          "banana", "apple", "mango", "orange"]

fruits_count = {}

for each in fruits:
    if each not in fruits_count :
        fruits_count[each] = 1
    else:
        fruits_count[each] += 1

print(fruits_count)
```

```python
logs = [
    "192.168.1.1 200 GET /home",
    "192.168.1.2 404 GET /about",
    "192.168.1.3 200 POST /login",
    "192.168.1.4 500 GET /data",
    "192.168.1.5 404 GET /contact",
    "192.168.1.6 200 GET /products",
    "192.168.1.7 500 POST /login",
    "192.168.1.8 403 GET /admin"
]

log_count = {}

for each_line in logs:
    content = each_line.split(" ")
    status_codes = content[1]

    if status_codes not in log_count:
        log_count[status_codes] = 1
    else:
        log_count[status_codes] += 1

print(log_count)
```

```python
import csv

with open("port_frequency.csv","w",newline="") as my_port:
    writer = csv.writer(my_port)

    for status_codes, code in log_count.items():
        writer.writerow([status_codes,code]) 
```
