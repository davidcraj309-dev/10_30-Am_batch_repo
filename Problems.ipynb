##### *72.Increasing Number Pattern*

```python
rows = int(input("Enter number of rows: "))

i = 1

while i<=rows:
     j = 1
     while j<=i:
          print(j,end="")
          j += 1

     i += 1
     print()
    
          
```

##### *73.Decreasing Number Pattern*

```python
rows = int(input("Enter number of rows: "))

i = 1

while i <= rows:
    j = 1

    while j <= rows - i + 1:
        print(j, end="")
        j += 1

    i += 1
    print()
```

##### *74.Decreasing Number Pattern*

```python
rows = int(input("Enter number of rows: "))

i = 1

while i<=rows:
    j = 1
    while j<=i:
        print("*",end="")
        j += 1

    i  += 1
    print()
```

##### *75.Decreasing Star Pattern*

```python
rows = int(input("Enter number of rows: "))

i = 1

while i <= rows:

    # Print spaces
    space = 1

    while space <= rows - i:
        print(" ", end="")
        space += 1

    # Print stars
    j = 1

    while j <= i:
        print("*", end="")
        j += 1

    i += 1
    print()
```
