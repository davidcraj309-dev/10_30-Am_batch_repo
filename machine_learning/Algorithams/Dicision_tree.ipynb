### **Important Decision Tree terminology**

| Term | Meaning |
|---|---|
| Root Node | Starting point of the tree |
| Decision Node | A question or condition |
| Branch | Possible path of a decision |
| Leaf Node | Final prediction |
| Splitting | Dividing data into groups |

Meaning: The starting point of a Decision Tree.
- It contains the first question.
- The algorithm starts making decisions from here.
Example:

Is the customer's monthly salary greater than ₹30,000?

### **2. Decision Node**
Meaning: A point where the algorithm asks a question or checks a condition.

Example:

Does the customer have a good credit score?
- Yes → Continue to the next decision.
- No → Loan rejected.

### **3. Branch**
Meaning: A path that represents the possible outcome of a decision.

Example:

If the question is:

Is salary greater than ₹30,000?

There are two branches:
- Yes
- No

### **4. Leaf Node**
Meaning: The final result or prediction made by the Decision Tree.

Example:
- Loan Approved
- Loan Rejected
There are no further decisions after reaching a leaf node.

### **5. Splitting**
Meaning: Dividing the dataset into smaller groups based on a condition.

Example:

Suppose we have 100 customer records.

The algorithm splits them based on salary:
- Salary greater than ₹30,000 → 60 customers.
- Salary less than or equal to ₹30,000 → 40 customers.
This process is called Splitting.

### **5.Model spiliting & Accuracy Training Stage**

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

input_dataset_train,input_dataset_test,output_dataset_train,output_dataset_test = train_test_split(input_dataset,output_dataset,test_size=0.2)
movie_model = DecisionTreeClassifier()
movie_model.fit(input_dataset_train,output_dataset_train) ## 80% trained
```

### **6.Model Test**

```python
movie_interest = movie_model.predict(input_dataset_test)
movie_interest
```

### **7.accuracy_score Checking**

```python

accuracy = accuracy_score(output_dataset_test,movie_interest) #actual output, model predicted values
accuracy
```
