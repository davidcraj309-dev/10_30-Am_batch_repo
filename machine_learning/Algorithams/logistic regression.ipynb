### **1. What is Gradient Descent?**

Gradient Descent is an optimization algorithm used in Machine Learning to find the best Weight and Bias by reducing the Loss Function.

Simple meaning:

The model makes predictions, calculates errors, and adjusts Weight and Bias to reduce those errors.

Real-world example:

We want to predict whether a student will pass or fail based on study hours.

Our goal is to train the model to predict the correct result.

### **2. Why do we need Gradient Descent?**

Initially, the model does not know the correct Weight and Bias.

It starts with initial values and gradually improves them.

Gradient Descent helps the model:
1. Make predictions.
2. Calculate prediction errors.
3. Calculate gradients.
4. Update Weight and Bias.
5. Reduce the Loss Function.
6. Repeat until the model learns.

### **Data**

```python
data = {
    "Age": [2.2,1.3,2.9,1.7,2.5],
    "admission": [1, 0, 1, 0, 1]
}

import pandas as pd

df = pd.DataFrame(data)
print(df)
```

```python
type(data)
```

```python
df
```

### **Model**

```python
from sklearn.linear_model import LinearRegression

regressor = LinearRegression()

X = df[["Age"]]
y = df["admission"]

regressor.fit(X, y)
```

```python
train_predict = regressor.predict(X)
train_predict
```

```python
plt.scatter(X,y)
plt.plot(X,train_predict,color="Red")
plt.xlabel("Age")
plt.ylabel("admission")
plt.show()
```

### **Percent converation.**

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
input_data = scaler.fit_transform(input_data)
print(input_data)
```

```python
input_data.shape
```

```python
output_data.shape
```

```python
from sklearn.model_selection import train_test_split

input_data_train,input_data_test,output_data_train,output_data_test = train_test_split(input_data,output_data,test_size=0.2)
```

```python
input_data_train.shape
```

### **logistic regression model**


```python
from sklearn.linear_model import LogisticRegression
from sklearn import metrics

model = LogisticRegression()
model.fit(input_data_train,output_data_train)
```

```python
predicted_survived = model.predict(input_data_test)
predicted_survived
```

### **confusion_matrix**

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(output_data_test,predicted_survived)
```

```python
pd.DataFrame(confusion_matrix(output_data_test,predicted_survived),columns=["Predicted No","Predicted Yes"],index=["Actual No","Actual yes"])
```

### **Accuracy Score**

```python
from sklearn.metrics import accuracy_score
accuracy_info = accuracy_score(output_data_test,predicted_survived)
```
