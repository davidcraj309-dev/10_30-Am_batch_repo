```python
customer_info["TotalCharges"] = pd.to_numeric(
    customer_info["TotalCharges"],
    errors="coerce"
)
```

```python
customer_info["TotalCharges"]  = customer_info["TotalCharges"].fillna(
    customer_info["TotalCharges"].median())
```

```python
X = customer_info.drop("Churn", axis=1)

y = customer_info["Churn"]
```

```python
X = pd.get_dummies(X, drop_first=True, dtype=int)
```

```python
y = y.map({
    "Yes": 1,
    "No": 0
})
```

```python
y.value_counts()
```

```python
print(X.shape)
print(X.dtypes)
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(
    X,
    y,
    test_size = 0.2,
)
```

### **Create your Decision Tree model**

```python
from sklearn.tree import DecisionTreeClassifier

churn_model = DecisionTreeClassifier(
    criterion="gini",
    max_depth=5,
    random_state=42
)

```

### **Train your model**

```python
churn_model.fit(x_train,y_train)
```

### **Make predictions**

```python
y_pred = churn_model.predict(x_test)
y_pred
```

### **Evaluate model accuracy**

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", accuracy)
```

### **Check the classification report**

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

### **Confusion matrix**

```python
from sklearn.metrics import confusion_matrix

print(confusion_matrix(y_test, y_pred))
```
