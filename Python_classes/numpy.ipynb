### **Numpy stands for numerical python**

* its a powerful library working with numbers
* every major data science library - pandas,Matplotlib,scikit-learn,tenserflow all of them built on the top numpy
* very faster than python
* provides      N-dimentional array                                                     

| Feature         | Python List                                        | NumPy Array                                         |
| --------------- | -------------------------------------------------- | --------------------------------------------------- |
| **Speed**       | Slow for numerical operations                      | Very fast                                           |
| **Memory**      | Uses more memory                                   | Uses less memory                                    |
| **Data Types**  | Can store different data types (**heterogeneous**) | Usually stores the same data type (**homogeneous**) |
| **Performance** | Slower for large numerical data                    | Faster for large numerical data                     |
| **Example**     | `[10, 20, 30]`                                     | `np.array([10, 20, 30])`                            |

### **speed comparison**

```python
import numpy as np
import time

start = time.time()

mylist = list(range(100000))
result = [x * 2 for x in mylist]
end = time.time()
print(f"python list time:{end-start:.4f} seconds")

start = time.time()

my_array = np.arange(100000)
result = my_array*2
end = time.time()

print(f"Numpy array time:{end-start:.4f} seconds")

```

```python
import numpy as np

#python list
py_list = [1,2,3,4,5]

#Numpy Array
np_arr = np.array([1,2,3,4,5])

#addition
print(py_list + [6,7,8,9,10]) # python list concotidanate into one list
print(np_arr + [6,7,8,9,10])    # numpy gives addition

#muliply
print(py_list*2)  #repeat!                   
print(np_arr*2)   # math                  

#mixed types
mixed_list = [1,"hello",55.77,90,True] ## combine all data types
print(mixed_list)

mixed_arr = np.array([1,"hello",55.77,90,True])### all become string
print(mixed_arr) 

#size in memory
import sys
print(sys.getsizeof(py_list)) ## larger size
print(np_arr.nbytes)  # smaller size - numpy wins
```
