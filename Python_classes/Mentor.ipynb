### 11.Python collection

**1.List** is a collection which is ordered and changeable. Allows duplicate members. Denoted by **Square Bracket[]**

* we should access index by sqaure bracket[0]

```python
mylist = ["hello", 27,"chennai","cat",20.55,20.55,True,True]
print(mylist,type(mylist))

print(mylist[5])

mylist[5] =  "Hi"
print(mylist)
```

**2.Tuple**  is a collection which is ordered and unchangeable. Allows duplicate members. Denoted by **() parentheses**

```python
mytuple =("hello", 27,"chennai","cat",20.55,20.55,True,True)
print(mytuple, type(mytuple))

print(mytuple[4])

mylist.append("david")
print(mylist)
```

**3.Set** is a collection which is unordered and unindexed. No duplicate members. Denoted by **{}- curly brackets**


```python
myset = {100,"hello","hi",100,"hello",True}
print(myset,type(myset))
```

### **Dictionary**
Dictionary is a collection that is unordered, changeable (mutable), and stores data in key-value pairs. Keys must be unique (no duplicate keys). It is denoted by curly braces {} using the format **{key: value}**.
Note: Python nested collections are also possible.

```python
mydict = {
    "name":"david",
    "age": 19,
    "city":"chennai",
    "Profession":"GenAi Engineer",
     10.10:90,
    "name":"yesu",
     True:False
     }

print(mydict,type(mydict))

print(mydict["name"])

mydict["age"] = 19
print(mydict)

```
