### **MUTABLE VS IMMUTABLE OBJECTS**

* Objects of built-in types like (int, float, bool, str, tuple) are immutable.
* Objects of built-in types like (list, set, dict) are mutable.

#### **Immutable**
* Same data stores in Same location
Different Data stores in Different location
* Existing value cannot be changed.
You can create a new value by reassignment.

#### **Mutable**
* Same data stores in Different location
Different Data stores in Different location

* Existing value can be changed.
No need to create a new object.

### **6.Python While loop** - *Condition statements*

```python
a = 1  # initialization / starting point

while a <=10: # comdition / Execution
    print(a, end = " ") # print all values one line
    a += 1  # incremention 

print("\nLoop is over") # string escape sequence \n - new line
```

### **break**

```python
a = 1

while a>0:
    if a == 5:
        print(a)
        break # code ends
    print(a,end=" ")
    a += 1
```

### **continue**

#### If the number is NOT divisible by 3, skip printing it.

```python
a = 1

while a<=20:
    if a%3 == 0:
        print(a,end=" ")
        a+= 1
        continue
    
    a += 1

print("\nloop over")
```

### **Nested while loop**

```python
rows = int(input("Enter rows: "))
columns = int(input("Enter columns: "))

a = 1

while a <= rows: # Rows
    b = 1
    
    while b <= columns: # Columns
        print(b, end="")
        b += 1

    a += 1
    print()
```

### **factorial**

```python
## factorial Funcation Automation

mylist1 = [1,2,3,4,5]
mylist2 = [1,2,3,4,5,6]
mylist3 = [1,2,3,4,5,6]


def my_factorial(mylist):
    x = 1

    for i in mylist:
       x = x*i
    return x 

ans  = my_factorial(mylist1)
print(ans)

print(my_factorial(mylist2))
print(my_factorial([1,2,3,4,5,6,7]))

```
