### **diferance between series & data frame**

| Feature    | **Series**                 | **DataFrame**                              |
| ---------- | -------------------------- | ------------------------------------------ |
| Dimension  | 1-Dimensional              | 2-Dimensional (multiple dim)                             |
| Structure  | Single column              | Multiple columns                           |
| Data       | One type of data column    | Multiple columns, different types          |
| Example    | `pd.Series([10, 20, 30])`  | `pd.DataFrame({"A":[10,20], "B":[30,40]})` |
| Main use   | Store one-dimensional data | Store tabular data                         |
| Similar to | One column in a table      | Excel table                                |

```python

import pandas as pd

df = pd.DataFrame({
    'Employee_ID': [101, 102, 103],
    'Employee_Name': ['Naveen', 'Arun', 'Midhun'],
    'Jan': [50000, 45000, 60000],
    'Feb': [55000, 47000, 62000],
    'Mar': [53000, 48000, 61000],
    'Apr': [58000, 50000, 65000],
    'May': [60000, 52000, 67000],
    'Jun': [62000, 55000, 70000]
})

print(df)
```

```python
df_long = pd.melt(
    df,
    id_vars=['Employee_ID', 'Employee_Name'],
    var_name='Month',
    value_name='Sales'
)

print(df_long)
```

```python
df_long.groupby("Month")["Sales"].sum()
```

```python
df_long.groupby("Employee_Name")["Sales"].sum()
```

```python

import seaborn as sns
import matplotlib.pyplot as plt

sns.lineplot(
    data=df_long,
    x="Month",
    y="Sales",
    hue="Employee_Name"
)

plt.show()
```
