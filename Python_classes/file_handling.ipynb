### **R** - Read mode
* default read mode

```python
f = open("notes.txt","r")  ## read mode
print(f.read())

```

```python
f = open("notes.txt")  ## read mode
print(f.read())

f.close()
```

### **open operation**

* if you open file must close the file using close()

### **W** - Write mode
* remove old conten adds new content

```python
file = open("notes.txt","w")
file.write("Hello Friends!\n")
file.write("welcome to python file Handling!\n")
file.write("This is Third Line!\n")

file.close()
```

### **Read operations**

```python
file = open("notes.txt","r")
line1 = file.read()
line2 = file.read()
print(line1)
print(line2)
```

### **Read operations**
* Readline reads first line
* create as list

```python
file = open("notes.txt","r")
lines =file.readline()
print(lines)

file.close()
```

### **Apend mode**
* adding new line in final

```python
file = open("notes.txt","a")
file.write("my fourth line")
file.close()
```

### **x-create mode**
* creating new file using **x**

```python
file_new = open("notes1.txt","x")
file_new.write("david Geni engineer\n")
file_new.write("need place in this month")

file_new.close()
```

### **with operation**

* if you open file no need use close()

```python
with open("notes.txt","r") as myfile:
    content = myfile.readline()
    print(content)
```

### **Delete File**
* os.remove()

```python
import os

os.remove("notes1.txt")
```

```python
if os.path.exists("notes1.txt"):
    os.remove("notes1.txt")
    print("the file is deleted")
else:
    print("File is not exist")
```
