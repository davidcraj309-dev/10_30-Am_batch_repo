#### **Python Decorators - Complete Notebook**

This notebook accompanies the lesson and demonstrates Python function superpowers, decorators, decorator factories, and multiple decorators using an ATM example.

#### **1. Superpower #1 - Functions are Objects**

```python
def withdraw_money(amount):
    print(f"💸 ₹{amount} withdrawn successfully.")

print(type(withdraw_money))
print(id(withdraw_money))
print(withdraw_money)
```

#### **2. Superpower #2 - Functions can be assigned to variables**

```python
atm_operation = withdraw_money

print(id(withdraw_money))
print(id(atm_operation))

atm_operation(5000)
```

#### **3. Superpower #3 - Functions can be passed as arguments**

```python
def process_transaction(operation, amount):
    print("🏧 Starting Transaction...")
    operation(amount)

process_transaction(withdraw_money, 3000)
```

#### **4. Superpower #4 - Functions can be returned from other functions**

```python
def atm_machine():
    def withdraw(amount):
        print(f"💸 ₹{amount} withdrawn successfully.")
    return withdraw

transaction = atm_machine()
transaction(4000)
```

#### **5. Superpower #5 - Closures**

```python
def atm_machine(pin):
    def withdraw(amount):
        print(f"🔐 PIN {pin} verified.")
        print(f"💸 ₹{amount} withdrawn successfully.")
    return withdraw

transaction = atm_machine(1234)
transaction(5000)
```

#### **6. Pain Point Without Decorators**

```python
def check_balance():
    print("🔐 Verifying ATM PIN...")
    print("💰 Available Balance: ₹50,000")

def deposit_money(amount, account_type):
    print("🔐 Verifying ATM PIN...")
    print(f"💵 ₹{amount} deposited into {account_type} account.")

def withdraw_money(amount, account_type):
    print("🔐 Verifying ATM PIN...")
    print(f"💸 ₹{amount} withdrawn from {account_type} account.")
```

#### **7. Basic Decorator**

```python
def verify_pin(func):
    def wrapper():
        print("🔐 Verifying ATM PIN...")
        func()
    return wrapper

@verify_pin
def check_balance():
    print("💰 Available Balance: ₹50,000")

check_balance()
```

#### **8. Problem - Function With Parameters**

```python
@verify_pin
def withdraw_money(amount):
    print(f"💸 ₹{amount} withdrawn successfully.")

# Uncomment to observe the error
withdraw_money(5000)
```

#### **9. Fix Using Wrapper Parameters**

```python
def verify_pin(func):
    def wrapper(amount):
        print("🔐 Verifying ATM PIN...")
        return func(amount)
    return wrapper

@verify_pin
def withdraw_money(amount):
    print(f"💸 ₹{amount} withdrawn successfully.")

withdraw_money(5000)
```

#### **10. Positional Arguments (args)**

```python
def verify_pin(func):
    def wrapper(*args):
        print("🔐 Verifying ATM PIN...")
        return func(*args)
    return wrapper

@verify_pin
def withdraw_money(amount, account_type):
    print(f"💸 ₹{amount} withdrawn from {account_type} account.")

withdraw_money(5000, "Savings")
```

#### **11. Keyword Arguments (kwargs)**

```python
def verify_pin(func):
    def wrapper(*args, **kwargs):
        print("🔐 Verifying ATM PIN...")
        return func(*args, **kwargs)
    return wrapper

@verify_pin
def withdraw_money(amount, account_type):
    print(f"💸 ₹{amount} withdrawn from {account_type} account.")

withdraw_money(amount=5000, account_type="Savings")
```

#### **12. Decorator Factory**

```python
def verify_pin(pin_length):
    def decorator(func):
        def wrapper(*args, **kwargs):
            print(f"🔐 Verifying {pin_length}-digit PIN...")
            return func(*args, **kwargs)
        return wrapper
    return decorator

@verify_pin(4)
def withdraw_money(amount):
    print(f"💸 ₹{amount} withdrawn successfully.")

withdraw_money(5000)
```

#### **13. Multiple Decorators**

```python
def verify_pin(func):
    def wrapper(*args, **kwargs):
        print("🔐 Verifying ATM PIN...")
        return func(*args, **kwargs)
    return wrapper

def check_balance(func):
    def wrapper(*args, **kwargs):
        print("💰 Checking Account Balance...")
        return func(*args, **kwargs)
    return wrapper

@verify_pin
@check_balance
def withdraw_money(amount, account_type):
    print(f"💸 ₹{amount} withdrawn from {account_type} account.")

withdraw_money(5000, "Savings")
```

#### **Summary**

* Functions are objects.
* Functions can be assigned to variables.
* Functions can be passed as arguments.
* Functions can be returned from other functions.
* Closures make decorators possible.
* Decorators eliminate duplicate code.
* *args handles positional arguments.
* **kwargs handles keyword arguments.
* Decorator factories allow decorators to accept arguments.
* Multiple decorators can be stacked.
