## **3.Python Collections**

* List is a collection which is ordered and changeable. Allows duplicate members. Denoted by []

* Tuple is a collection which is ordered and unchangeable. Allows duplicate members. Denoted by ()

* Set is a collection which is unordered and unindexed. No duplicate members. Denoted by {}

* Dictionary is a collection that is unordered, changeable (mutable), and stores data in key-value pairs. Keys must be unique (no duplicate keys). It is denoted by curly braces {} using the format {key: value}.

### **python for loop**

For loop is used for iterating over a sequence (that is either a list, a tuple, a dictionary, a set, or a string). With the for loop we can execute a set of statements, once for each item in a list, tuple, set etc.

Keyword

### **for**

Note: range() – is a Python standard function used for numbers.

### syntax

* **for \<variable_name>\ in <str/list/tuple/set/dict>:**


```python
for x in "hello world":
    print(x,end="")
```

```python
mystr = "hello world" ## 

for x in mystr: ## end before 0:5 ## fetechs letter by letter
    print(x,end="")
```

```python
## list collections

mylist = [10,20.55,"hello",True,20.55,"hello"] ## Items

for x in mylist:
    print(x,end=",")
```

```python
## tuple collections

mylist = (10,20.55,"hello",True,20.55,"hello")

for x in mylist:
    print(x,end=",")
```

```python
mylist = [2,6,5,11,16,17]

for each in mylist:
    if each%5 == 0:
        print(each,"Divisiable Buy 5")
        break

else:
 print("number divisiable not found")
```
