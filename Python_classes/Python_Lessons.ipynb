## **4.Python Collections**

* List is a collection which is ordered and changeable. Allows duplicate members. Denoted by []

* Tuple is a collection which is ordered and unchangeable. Allows duplicate members. Denoted by ()

* Set is a collection which is unordered and unindexed. No duplicate members. Denoted by {}

### **RECURSION**

A function is called within the same function is called Recursion.

Recursion is a common mathematical and programming concept. It means that a function calls itself. This has the benefit of meaning that you can loop through data to reach a result.

The developer should be very careful with recursion as it can be quite easy to slip into writing a function which never terminates, or one that uses excess amounts of memory or processor power. However, when written correctly recursion can be a very efficient and mathematically-elegant approach to programming.

````python
def myadd(x,y):
    return x+y
def mysub(x,y):
    return x-y

print(mysub(myadd(10,20),myadd(5,10)))
````

````python
def myfactorial(num):
    pro = 1

    for i in range(1,num+1):
        pro = pro*i

    return pro

ans  = myfactorial(5)
print(ans)
````

````python
def myrec_factorial(num):
    if num == 1:
        return 1
    else:
        return num * myrec_factorial(num-1)
ans = myrec_factorial(5)
print(ans)
````

## REGEX META CHARACTERS

- `[]` → A set of characters
  - Example: `[a-m]`

- `\` → Signals a special sequence / escapes special characters
  - Example: `\d`

- `.` → Any character except newline
  - Example: `he..o`

- `^` → Starts with
  - Example: `^hello`

- `$` → Ends with
  - Example: `world$`

- `*` → Zero or more occurrences
  - Example: `aix*`

- `+` → One or more occurrences
  - Example: `aix+`

- `?` → Zero or one occurrence
  - Example: `aix?`

- `{}` → Exactly the specified number of occurrences
  - Example: `al{2}`

- `|` → Either / OR
  - Example: `falls|stays`

- `()` → Capture and group
  - Example: `(aix)+`

# REGEX : SETS

| Set | Description |
|---|---|
| `[arn]` | Returns a match where one of the specified characters (a or r or n) are present |
| `[a-n]` | Returns a match for any lower case character, alphabetically between a and n |
| `[^arn]` | Returns a match for any character EXCEPT a, r, and n |
| `[0123]` | Returns a match where any of the specified digits (0, 1, 2, or 3) are present |
| `[0-9]` | Returns a match for any digit between 0 and 9 |
| `[0-5][0-9]` | Returns a match for any two-digit numbers from 00 and 59 |
| `[a-zA-Z]` | Returns a match for any character alphabetically between a and z, lower case OR upper case |
| `[+]` | In sets, `+`, `*`, `.`, `|`, `()`, `{}`, `$` has no special meaning, so `[+]` means: return a match for any `+` character in the string |

## REGEX : SPECIAL SEQUENCES

- `\A` → Match if characters are at the beginning of the string
  - Example: `\AThe`

- `\b` → Match at the beginning or end of a word
  - Example: `r"\bain"`
  - Example: `r"ain\b"`

- `\B` → Match when characters are NOT at the beginning or end of a word
  - Example: `r"\Bain"`
  - Example: `r"ain\B"`

- `\d` → Matches digits (0-9)
  - Example: `\d`

- `\D` → Matches characters that are NOT digits
  - Example: `\D`

- `\s` → Matches whitespace characters
  - Example: `\s`

- `\S` → Matches characters that are NOT whitespace
  - Example: `\S`

- `\w` → Matches word characters (a-z, A-Z, 0-9, `_`)
  - Example: `\w`

- `\W` → Matches characters that are NOT word characters
  - Example: `\W`

- `\Z` → Match if characters are at the end of the string
  - Example: `Spain\Z`

## REGEX FUNCTIONS

| Function | Description |
|----------|-------------|
| `findall()` | Returns a list containing all matches |
| `search()` | Returns a Match object if there is a match anywhere in the string |
| `split()` | Returns a list where the string has been split at each match |
| `sub()` | Replaces one or many matches with a string |

````python
with open("myfile.txt","r") as file:
    for line in file:
        print(line.strip())
````

### **login user feedback**

````python
feedback = input("enter your feedback:")

with open("myfile.txt","a") as log:
    log.write(feedback + "\n")

print("Thanks for your feedback")
````

````python
with open("myfile.txt","r") as file:
    print(file.readline().strip())
    print(file.readline().strip())
    print(file.readline().strip())
````

### **5.readline()**

````python
with open("file.txt") as f:
    while True:
        line = f.readline()

        if not line:
            break
        if "ERROR" in line:
            print("Found error:", line.strip())
````
