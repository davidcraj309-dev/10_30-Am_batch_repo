```python
import matplotlib.pyplot as plt

month = ["jan","Feb","Mar","Apr","May"]
sales = [12000,15000,11000,18000,22000]

plt.plot(month,sales,color ="red",marker = "o",linestyle="-") # => month - X Axis || sales - Y Axis
plt.title("Monthly sales")
plt.xlabel("Months")
plt.grid()
plt.ylabel("Sales")
plt.show();
```

### **1.Figure and Axes**

```python
import matplotlib.pyplot as plt

month = ["jan","Feb","Mar","Apr","May"]
sales = [12000,15000,11000,18000,22000]


plt.plot(month,sales)
```

```python
fig, ax = plt.subplots()

ax.plot(month,sales)
plt.show();
```

### **Figure size and layout**

```python
fig , ax  = plt.subplots(figsize=(10,5)) #width ,#hieght
ax.plot(month,sales)
plt.tight_layout()
plt.show();
```

### **Titles,Labels,and Ticks**

```python
fig, ax = plt.subplots(figsize=(10,5))

ax.plot(month,sales)

ax.set_title("Monthly sales perfomernce - 2026", fontsize=16,fontweight="bold")
ax.set_xlabel("month",fontsize=16,fontweight="bold")
ax.set_ylabel("sales",fontsize=16,fontweight="bold")
plt.tight_layout()

```

### **Lines,Markers,colour and grids**

```python

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(
    month,
    sales,
    color="#5D09ED",
    linewidth=3,
    linestyle="-",
    marker="o",
    markersize=10,
    label="Sales"
)

ax.set_title(
    "Monthly Sales Performance - 2026",
    fontsize=16,
    fontweight="bold"
)
ax.set_xlabel("Month")
ax.set_ylabel("Sales Amount ($)")

plt.grid(axis="y",alpha =0.5)
ax.legend()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv('sales_data_practice.csv')
df.head()
```

```python
df["Revenue"].plot(kind="line")
plt.show();
```

```python
df.info()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df.info()
```

```python
monthly = df.groupby(df['Date'].dt.month)["Revenue"].sum()
print(monthly)
```

```python
monthly.plot(kind='line',marker="o",figsize=(10,5),color="Red")

plt.title("Monthly Revenue Trend",fontsize=16,fontweight="bold")
plt.xlabel("Month")
plt.ylabel("Revenue($)")
```

### **which category high revenue generated to check (use pie chart)**

```python
category_sales = df.groupby("Category")["Revenue"].sum().sort_values(ascending=True)
print(category_sales)
```

```python
category_sales.plot(kind='bar',figsize=(9,5),color="green")
plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Revenue($)")
plt.xticks(rotation= 0)

plt.tight_layout()
plt.show();
```

### **stack barchart**

```python
region_category = df.groupby(["Region","Category"])["Revenue"].sum().unstack()
print(region_category)
```

```python
region_category.plot(kind="bar",figsize=(11,5),stacked=True)

plt.title("Revenue by Region and Category")
plt.xlabel("Region")
plt.ylabel("Revenue($)")
plt.legend(title="Category")
plt.xticks(rotation= 0)

plt.tight_layout()
plt.show();
```

```python
region_share = df["Region"].value_counts()
print(region_share)
```

```python
region_share.plot(
    kind="pie",
    autopct="%1.0f%%",
    figsize=[6,6]              
                  
)

plt.title("Orders by Region")
plt.tight_layout()
plt.show();
```
