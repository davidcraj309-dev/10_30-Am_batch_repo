````python
# Mission 60 — Document Cleaner
text = """
   Python is powerful.

   Python is useful for GenAI.
"""

lines = text.splitlines()

clean_lines = []

for line in lines:
    line = line.strip()
    
    if line:
        clean_lines.append(line)

result = "\n".join(clean_lines)

print(result)
````

### **👑 FINAL BOSS — GENAI TEXT ENGINE**

````python
text = """
   I am David.

   I am learning PYTHON, RAG, and LLM.

   Python is powerful.

   I want to become a GenAI Engineer.
"""

# 1. Clean the text
text = text.strip()

# 2. Process the lines
lines = text.splitlines()

# 3. Remove empty lines
clean_lines = []

for line in lines:
    line = line.strip()
    
    if line:
        clean_lines.append(line)

# 4. Extract the skills
skills = []

for line in clean_lines:
    if "PYTHON" in line.upper():
        skills.append("Python")
    
    if "RAG" in line.upper():
        skills.append("RAG")
    
    if "LLM" in line.upper():
        skills.append("LLM")

# Remove duplicate skills
skills = list(dict.fromkeys(skills))

# 5. Count "Python"
clean_text = "\n".join(clean_lines)

python_count = clean_text.lower().count("python")

# 6. Check whether GenAI exists
genai_exists = "genai" in clean_text.lower()

# 7. Find the longest line
longest_line = max(clean_lines, key=len)

# 8. Count characters
character_count = len(clean_text)

# 9. Create clean final string
final_string = "\n".join(clean_lines)

print("----- RESULTS -----")
print("Skills:", skills)
print("Python count:", python_count)
print("GenAI exists:", genai_exists)
print("Longest line:", longest_line)
print("Character count:", character_count)
print("\n----- CLEAN TEXT -----")
print(final_string)
````
