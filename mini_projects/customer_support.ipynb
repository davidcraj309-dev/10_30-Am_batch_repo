```python
print("==========================")
print("CUSTOMER SUPPORT SYSTEM")
print("===========================")

print()

class Customer:
    def __init__(self,name,email,customer_id):
        self.name = name
        self.email = email
        self.customer_id = customer_id

    def display_customer(self):
       
           print("Customer Details")
           print("-----------------")

           print(f"name: {self.name}")
           print(f"email: {self.email}")
           print(f"customer_id: {self.customer_id}")

class Ticket:
    def __init__(self,Ticket_id,Issue,Type,Priority,Status):
        self.Ticket_id  = Ticket_id
        self.Issue = Issue
        self.Type = Type
        self.Priority = Priority
        self.Status = Status

    def display_ticket(self):
           print("ticket Details")
           print("-----------------")

           print(f"Ticket_id: {self.Ticket_id}")
           print(f"Issue: {self.Issue}")
           print(f"Type: {self.Type}")
           print(f"Priority: {self.Priority}")
           print(f"Status: {self.Status}")

class Agent:
    def __init__(self,Agent_id,name,Specialization):
        self.Agent_id  = Agent_id
        self.name = name
        self.Specialization = Specialization

    def display_agent(self):
        print("Agent Details")
        print("-----------------")
        print(f"Agent_id: {self.Agent_id}")
        print(f"name: {self.name}")
        print(f"Specialization: {self.Specialization}")
        

# Customer object

Customer1 = Customer("David", "david@gmail.com", 144)
Customer1.display_customer()

print()

# Ticket object

Ticket1 = Ticket(
     "T001",
     "Payment failed",
     "Billing",
     "High",
     "Open"
)

Ticket1.display_ticket()

print()

# Agent object

Agent1 = Agent(309,"melvin","Billing")
Agent1.display_agent()


```
