###  **\------TREASURE HUNT GAME\------------**

```python
print("=============================")
print("     TREASURE HUNT GAME   ")
print("=============================")

score = 100
choice = int(input("choose a door (1,2,3):"))

if choice == 1:
    print("Monster!")
    score = score - 30
    print("your score is decreased by 30:",score)
    fight = int(input("Fight the monster? (1 = Yes, 2 = No): "))

    if fight == 1:
        score = score + 50
        print("your score is increased by 50 :",score)
        print("You defeated the monster!")
    elif fight == 2:
        score = score - 20
        print("your score is decreased by 20 :",score)
        print("You ran away!")

    else:
        print("Invalid choice!")

elif choice == 2:
    print("You found the treasure! ")
    score = score + 50
    print("your score is increased by 50 :",score)
    box = int(input("open the box? (1=Yes,2=No)"))

    if box == 1:
        score = score + 50
        print("your score is increased by 50 :",score)
        print("You found a diamond!")
    elif box == 2:
        print("You left the treasure!")

    else:
      print("Invalid choice!")

elif choice == 3:
    print("Trap!")
    score = score - 40
    print("your score is decreased by 40 :",score)

    if score <= 50:
        print("game over!")
    else:
        print("You escaped the trap!")

else:
 print("Invalid choice!")

print("Final score:", score)
```
